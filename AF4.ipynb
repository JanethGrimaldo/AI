```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img =cv2.imread("ave_2.png",0) # Carga de imagen en escala de grises
alto, ancho = img.shape

Matriz = np.float32([[1,0,50], [0,1,30]]) # Desplazamiento
Traslacion = cv2.warpAffine(img, Matriz, (ancho, alto))

Escalado = cv2.resize(img, (0, 0), fx=2, fy=2) # Escalado

#Rotación
centro = (ancho//2, alto//2) #Uso de división entera //
Matriz_rotacion = cv2.getRotationMatrix2D(centro, 50,1.5)
rotacion = cv2.warpAffine(img, Matriz_rotacion,(ancho,alto))

reflexion = cv2.flip(img, 1) #Reflexión

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten() # Convertimos la matriz 2D de subplots en un array 1D

axes[0].imshow(img, cmap="gray")
axes[0].set_title("Original")
axes[0].axis("off")

axes[1].imshow(Traslacion, cmap="gray")
axes[1].set_title("Traslación")
axes[1].axis("off")

axes[2].imshow(Escalado, cmap="gray")
axes[2].set_title("Escalado")
axes[2].axis("off")

axes[3].imshow(rotacion, cmap="gray")
axes[3].set_title("Rotacion con Escalado (1.5)")
axes[3].axis("off")

axes[4].imshow(reflexion, cmap="gray")
axes[4].set_title("Reflexión")
axes[4].axis("off")

axes[5].axis("off")

plt.tight_layout()
plt.show()
```
