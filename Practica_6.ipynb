**Librerías necesarias**

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

**Carga de ambas imágenes**

```python
img1 = cv2.imread("1.jpg", cv2.IMREAD_GRAYSCALE)
img2 = cv2.imread("2.jpg", cv2.IMREAD_GRAYSCALE)
```

**Redimensionamiento**

Para asegurarnos que ambas imágenes tienen el mismo tamaño se utiliza cv2.resize()

```python

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
```

Se aplica la suma ponderada mediante la función cv2.addWeighted(), la cual combina ambas imágenes asignando un peso específico a cada una. Los parámetros alpha; que corresponde a la imagen no. 1, y beta; que corresponde a la imagen no. 2, controlan la proporción de cada imagen en el resultado final, permitiendo resaltar una sobre la otra.

```python
alpha = 0.6
beta = 0.4

suma_pond = cv2.addWeighted(img1, alpha, img2, beta, 0)
```

La suma promedio consiste en sumar ambas imágenes y dividir el resultado entre dos, se utiliza astype(np.uint8) para asegurar que los valores se mantengan dentro del rango válido de intensidades que es de 0 a 255.

```python
suma_prom = (img1 + img2) / 2
suma_prom = suma_prom.astype(np.uint8)
```

Resultados:

```python
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.imshow(img1, cmap="gray")
plt.title("Imagen 1")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(img2, cmap="gray")
plt.title("Imagen 2")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(suma_pond, cmap="gray")
plt.title("Suma ponderada")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(suma_prom, cmap="gray")
plt.title("Suma promedio")
plt.axis("off")

plt.tight_layout()
plt.show()
```

La suma ponderada permitió combinar ambas imágenes controlando la contribución de cada una mediante los pesos establecidos. Como resultado, se obtuvo una imagen más equilibrada, donde se redujo el ruido gracias a la versión suavizada, mientras que se conservaron los detalles estructurales provenientes de la imagen con mayor contraste.

```python
img3 = cv2.imread("3.jpg", cv2.IMREAD_GRAYSCALE)
img4 = cv2.imread("4.jpg", cv2.IMREAD_GRAYSCALE)

img4 = cv2.resize(img4, (img3.shape[1], img3.shape[0]))
```

**Resta de imágenes**

Esta operación puede generar valores negativos para evitar este problema. La función garantiza que todos los valores sean positivos, permitiendo una representación correcta de las diferencias entre las imágenes y resaltando las zonas donde existen cambios de intensidad.

```python
resta = img3 - img4
resta_abs = cv2.absdiff(img3, img4)
```

Resultados:

```python
plt.figure(figsize=(12,6))

plt.subplot(2,2,1)
plt.imshow(img3, cmap="gray")
plt.title("3")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(img4, cmap="gray")
plt.title("4")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(resta, cmap="gray")
plt.title("Resta directa")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(resta_abs, cmap="gray")
plt.title("Resta absoluta")
plt.axis("off")

plt.tight_layout()
plt.show()
```

**Multiplicación por un factor**

Se definieron tres factores de multiplicación: 1.5, 2 y 5.

```python
fact1=1.5
fact2=2
fact3=5
```

Se realizó la multiplicación de cada imagen por los factores definidos.

```python
img1_mult1=cv2.multiply(img1, np.array([fact1], dtype=np.float32))
img1_mult2=cv2.multiply(img1, np.array([fact2], dtype=np.float32))
img1_mult3=cv2.multiply(img1, np.array([fact3], dtype=np.float32))

img3_mult1=cv2.multiply(img3, np.array([fact1], dtype=np.float32))
img3_mult2=cv2.multiply(img3, np.array([fact2], dtype=np.float32))
img3_mult3=cv2.multiply(img3, np.array([fact3], dtype=np.float32))
```

Se realizó la conversión de las imágenes resultantes a formato uint8. Esta función permite ajustar los valores de intensidad al rango válido de visualización.

```python
img1_mul1=cv2.convertScaleAbs(img1_mult1)
img1_mul2=cv2.convertScaleAbs(img1_mult2)
img1_mul3=cv2.convertScaleAbs(img1_mult3)

img3_mul1=cv2.convertScaleAbs(img3_mult1)
img3_mul2=cv2.convertScaleAbs(img3_mult2)
img3_mul3=cv2.convertScaleAbs(img3_mult3)
```

Resultados:

```python
plt.figure(figsize=(10, 4))

plt.subplot(2,4,1)
plt.imshow(img1, cmap='gray')
plt.title("Imagen Original 1")
plt.axis("off")

plt.subplot(2,4,5)
plt.imshow(img3, cmap='gray')
plt.title("Imagen Original 2")
plt.axis("off")

plt.subplot(2,4,2)
plt.imshow(img1_mult1, cmap='gray')
plt.title(f"Imagen 1x {fact1}")
plt.axis("off")
plt.subplot(2,4,3)
plt.imshow(img1_mult2, cmap='gray')
plt.title(f"Imagen 1x {fact2}")
plt.axis("off")
plt.subplot(2,4,4)
plt.imshow(img1_mult3, cmap='gray')
plt.title(f"Imagen 1x {fact3}")
plt.axis("off")

plt.subplot(2,4,6)
plt.imshow(img3_mult1, cmap='gray')
plt.title(f"Imagen 2x {fact1}")
plt.axis("off")
plt.subplot(2,4,7)
plt.imshow(img3_mult2, cmap='gray')
plt.title(f"Imagen 2x {fact2}")
plt.axis("off")
plt.subplot(2,4,8)
plt.imshow(img3_mult3, cmap='gray')
plt.title(f"Imagen 2x {fact3}")
plt.axis("off")

plt.show()
```

**Multiplicación elemento por elemento**

```python
img5 = cv2.imread("5.jpg", cv2.IMREAD_GRAYSCALE)
img6 = cv2.imread("6.jpg", cv2.IMREAD_GRAYSCALE)

img6 = cv2.resize(img6, (img5.shape[1], img5.shape[0]))
```

La multiplicación píxel a píxel utilizando la función cv2.multiply().

```python
img_multiplicada=cv2.convertScaleAbs(img_multiplicada)
```
